```python
import pandas as pd
from decimal import Decimal
from IPython.display import display

# Run this only if users_data is not already loaded
users_data = pd.read_csv('/content/users_data.csv')


# Required column check

required_columns = [
    "current_age",
    "retirement_age",
    "birth_year",
    "birth_month",
    "latitude",
    "longitude",
    "credit_score",
    "num_credit_cards",
    "gender",
    "per_capita_income",
    "yearly_income",
    "total_debt"
]

missing_required_columns = [
    column
    for column in required_columns
    if column not in users_data.columns
]

if missing_required_columns:
    raise KeyError(
        f"Required columns not found: {missing_required_columns}"
    )

row_count, column_count = users_data.shape

if row_count == 0:
    raise ValueError("users_data contains no rows.")


# Exact percentage function

def exact_percentage(count, total):
    percentage = (
        Decimal(int(count)) * Decimal("100")
    ) / Decimal(int(total))

    return f"{format(percentage, 'f')}%"


# A. Structure

structure_profile = pd.DataFrame({
    "Metric": [
        "Rows",
        "Columns",
        "Memory usage in bytes"
    ],
    "Value": [
        row_count,
        column_count,
        int(
            users_data.memory_usage(
                index=True,
                deep=True
            ).sum()
        )
    ]
})

print("A. STRUCTURE")
display(structure_profile)


# B. Missingness for every column

null_counts = users_data.isna().sum()

missingness_profile = pd.DataFrame({
    "Column": users_data.columns,
    "Null count": null_counts.to_numpy(),
    "Null percentage": [
        exact_percentage(count, row_count)
        for count in null_counts.to_numpy()
    ]
})

print("\nB. MISSINGNESS")
display(missingness_profile)


# C. Cardinality for every column
# Null values are excluded because they are reported separately above.

distinct_counts = users_data.nunique(dropna=True)

cardinality_profile = pd.DataFrame({
    "Column": users_data.columns,
    "Distinct values": distinct_counts.to_numpy()
})

print("\nC. CARDINALITY")
display(cardinality_profile)


# D. Numerical ranges

numerical_fields = [
    "current_age",
    "retirement_age",
    "birth_year",
    "birth_month",
    "latitude",
    "longitude",
    "credit_score",
    "num_credit_cards"
]

non_numeric_fields = [
    column
    for column in numerical_fields
    if not pd.api.types.is_numeric_dtype(users_data[column])
]

if non_numeric_fields:
    raise TypeError(
        f"Expected numerical columns are not numerical: "
        f"{non_numeric_fields}"
    )

numerical_ranges_profile = (
    users_data[numerical_fields]
    .agg(["min", "max", "mean", "median"])
    .T
    .rename(columns={
        "min": "Minimum",
        "max": "Maximum",
        "mean": "Mean",
        "median": "Median"
    })
    .reset_index()
    .rename(columns={"index": "Column"})
)

print("\nD. NUMERICAL RANGES")
display(numerical_ranges_profile)


# E. Gender categories

gender_counts = users_data["gender"].value_counts(
    dropna=False,
    sort=False
)

gender_profile = (
    gender_counts
    .rename_axis("Category")
    .reset_index(name="Exact count")
)

gender_profile["Exact percentage"] = [
    exact_percentage(count, row_count)
    for count in gender_profile["Exact count"]
]

print("\nE. GENDER CATEGORIES")
display(gender_profile)


# F. Raw financial fields
# No cleaning, symbol removal or datatype conversion is performed.

financial_fields = [
    "per_capita_income",
    "yearly_income",
    "total_debt"
]

raw_financial_profile = pd.DataFrame({
    "Column": financial_fields,
    "Datatype": [
        str(users_data[column].dtype)
        for column in financial_fields
    ],
    "First 5 values": [
        users_data[column].head(5).tolist()
        for column in financial_fields
    ]
})

print("\nF. RAW FINANCIAL FIELDS")

with pd.option_context("display.max_colwidth", None):
    display(raw_financial_profile)
```

### TRANSACTIONS - Lifecycle Anomalies

```python
import pandas as pd

# Ensure df and cards_df are loaded
transactions_df = pd.read_csv('/content/transactions_data.csv')
cards_df = pd.read_csv('/content/cards_data.csv')

# Corrected explicit date parsing
cards_df["acct_open_parsed"] = pd.to_datetime(
    cards_df["acct_open_date"],
    format="%m/%Y",
    errors="coerce"
)

cards_df["expires_parsed"] = pd.to_datetime(
    cards_df["expires"],
    format="%m/%Y",
    errors="coerce"
)

transactions_df["date_parsed"] = pd.to_datetime(
    transactions_df["date"],
    format="%Y-%m-%d %H:%M:%S",
    errors="coerce"
)

# Merge transactions with card data to get open and expiry dates
transactions_with_card_info = pd.merge(transactions_df, cards_df[['id', 'acct_open_parsed', 'expires_parsed', 'card_type', 'num_cards_issued']],
                                     left_on='card_id', right_on='id', suffixes=('_trans', '_card'))

# Compare month to month
transaction_month = transactions_with_card_info["date_parsed"].dt.to_period("M")
opening_month = transactions_with_card_info["acct_open_parsed"].dt.to_period("M")
expiry_month = transactions_with_card_info["expires_parsed"].dt.to_period("M")

before_open = transaction_month < opening_month
after_expiry = transaction_month > expiry_month

print(f"\nTransactions before opening month: {before_open.sum()} rows")
print(f"Distinct cards involved: {transactions_with_card_info.loc[before_open, 'card_id'].nunique()}")

print(f"\nTransactions after expiry month: {after_expiry.sum()} rows")
print(f"Distinct cards involved: {transactions_with_card_info.loc[after_expiry, 'card_id'].nunique()}")

# Distribution of card_type and num_cards_issued for anomalous cards (retaining previous logic)
anomalous_cards = transactions_with_card_info.loc[before_open | after_expiry, 'card_id'].unique()
anomalous_cards_df = cards_df[cards_df['id'].isin(anomalous_cards)]

print("\nDistribution of card_type for anomalous cards:")
display(anomalous_cards_df['card_type'].value_counts())

print("\nDistribution of num_cards_issued for anomalous cards:")
display(anomalous_cards_df['num_cards_issued'].value_counts())
```

## Investigation 2: Lifecycle anomaly severity

```python
import numpy as np

# Re-load or ensure the necessary DataFrames and Series from cell 9bd07a57 are available.
# This ensures the context for lifecycle anomalies is fresh and accurate.

transactions_df = pd.read_csv('/content/transactions_data.csv')
cards_df = pd.read_csv('/content/cards_data.csv')

cards_df["acct_open_parsed"] = pd.to_datetime(
    cards_df["acct_open_date"],
    format="%m/%Y",
    errors="coerce"
)

cards_df["expires_parsed"] = pd.to_datetime(
    cards_df["expires"],
    format="%m/%Y",
    errors="coerce"
)

transactions_df["date_parsed"] = pd.to_datetime(
    transactions_df["date"],
    format="%Y-%m-%d %H:%M:%S",
    errors="coerce"
)

transactions_with_card_info = pd.merge(transactions_df, cards_df[['id', 'acct_open_parsed', 'expires_parsed', 'card_type', 'num_cards_issued']],
                                     left_on='card_id', right_on='id', suffixes=('_trans', '_card'))

transaction_month = transactions_with_card_info["date_parsed"].dt.to_period("M")
opening_month = transactions_with_card_info["acct_open_parsed"].dt.to_period("M")
expiry_month = transactions_with_card_info["expires_parsed"].dt.to_period("M")

before_open = transaction_month < opening_month
after_expiry = transaction_month > expiry_month

print("### A. For transactions before opening")
anomalous_before_open_df = transactions_with_card_info[before_open].copy()
anomalous_before_open_df['months_before_opening'] = (opening_month[before_open] - transaction_month[before_open]).apply(lambda x: x.n)

print("Months before opening:")
print(f"  - Minimum: {anomalous_before_open_df['months_before_opening'].min()} months")
print(f"  - Median: {anomalous_before_open_df['months_before_opening'].median()} months")
print(f"  - Maximum: {anomalous_before_open_df['months_before_opening'].max()} months")

print("\nCounts for months before opening:")
bins = [0, 1, 6, 12, np.inf]
labels = ['1 month before', '2 to 6 months before', '7 to 12 months before', 'more than 12 months before']
months_before_counts = pd.cut(anomalous_before_open_df['months_before_opening'], bins=bins, labels=labels, right=True).value_counts().sort_index()
display(months_before_counts)
```

```python
print("### B. For transactions after expiry")
anomalous_after_expiry_df = transactions_with_card_info[after_expiry].copy()
anomalous_after_expiry_df['months_after_expiry'] = (transaction_month[after_expiry] - expiry_month[after_expiry]).apply(lambda x: x.n)

print("Months after expiry:")
print(f"  - Minimum: {anomalous_after_expiry_df['months_after_expiry'].min()} months")
print(f"  - Median: {anomalous_after_expiry_df['months_after_expiry'].median()} months")
print(f"  - Maximum: {anomalous_after_expiry_df['months_after_expiry'].max()} months")

print("\nCounts for months after expiry:")
bins = [0, 1, 6, 12, np.inf]
labels = ['1 month after', '2 to 6 months after', '7 to 12 months after', 'more than 12 months after']
months_after_counts = pd.cut(anomalous_after_expiry_df['months_after_expiry'], bins=bins, labels=labels, right=True).value_counts().sort_index()
display(months_after_counts)
```
